```python
import random
from datetime import datetime, timedelta

# Function to generate a random date within the last 20 years
def generate_random_date():
    # Define the start and end date
    end_date = datetime.now()
    start_date = end_date - timedelta(days=20 * 365)

    # Generate a random date
    random_date = start_date + timedelta(days=random.randint(0, 20 * 365))
    return random_date.strftime('%Y%m%d')

def generate_random_name():
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)
    return f"{first_name} {last_name}"
```

```python
# pre-tagging for creating historical data 
def getPossibleColumnList( colNameToCheck):
    match colNameToCheck:
        case "name":
            return possible_name_columns
        case "email":
            return possible_mail_columns
        case _:
            return possible_country_columns

def getPossibleColumns(colNameToCheck,df):
    for col in getPossibleColumnList(colNameToCheck):
        if col in df.columns:
            name_column = col
            break
    return name_column

def derivedNewColumns(df):
    countryCol= getPossibleColumns("country",df)
    nameCol = getPossibleColumns("name",df)
    mailCol=  getPossibleColumns("email",df)
    df["DerivedName"]=df[nameCol]
    df["DerivedMail"]=df[mailCol]
    df["DerivedCountry"]=df[countryCol]
    df['DerivedCountryCode'] = df[countryCol].map(country_codes)
    return df

def applyStaticRules(ndf):
    df = derivedNewColumns(ndf)
    df["is_pii"] =  (df['DerivedName'].isna() | (df['DerivedName'] == '')) &(df['DerivedMail'].isna() | (df['DerivedMail'] == '')) & (df['ssn'].isna() | (df['ssn'] == '')) |  ((df['DerivedName'] == 'restricted') & (df['DerivedMail'] == 'restricted') & (df['ssn'] == 'restricted'))

    df["is_gdpr"] = df["DerivedCountryCode"].isin(gdpr_country_codes)
    df["is_geoRestricted"] = df['DerivedCountryCode'].isin(cloud_restriction)
    return df   
```
